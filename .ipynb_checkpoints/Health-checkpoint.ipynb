```python
!pip install -q git+https://github.com/tensorflow/docs
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import tensorflow as tf
try:
  %tensorflow_version 2.x
except Exception:
  pass
from tensorflow import keras
from tensorflow.keras import layers

import tensorflow_docs as tfdocs
import tensorflow_docs.plots
import tensorflow_docs.modeling
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
# Import data
df = pd.read_csv('insurance.csv')
print(df.head())
print(df.size)
```

```python
label_Encoder = LabelEncoder()
for column in df.columns:
    if df[column].dtype == 'object':
        if len(df[column].unique()) <= 4:
            df[column] = label_Encoder.fit_transform(df[column])
        else:
            d = pd.get_dummies(df[column])
            df = pd.concat([df, d], axis=1)
            df = df.drop(column, axis=1)
```

```python
shuffled_df= df.sample(frac=1, random_state=42)
train_df, test_df = train_test_split(shuffled_df, test_size=0.2, random_state=42)

train_df = train_df.reset_index(drop=True)
test_df = test_df.reset_index(drop=True)

train_labels = train_df.pop('expenses')
test_labels = test_df.pop('expenses')
```

```python
normalizer = layers.Normalization()
normalizer.adapt(np.array(train_df))

model = keras.Sequential([
    normalizer,
    layers.Dense(16),
    layers.Dense(4),
    layers.Dropout(.2),
    layers.Dense(1),
])
```

```python
model.compile(
    optimizer=tf.optimizers.Adam(learning_rate=0.1),
    loss='mae',
    metrics=['mae', 'mse']
)
model.build()
model.summary()
```

```python
history = model.fit(
    train_df,
    train_labels,
    epochs=100,
    validation_split=0.5,
    verbose=0, 
)

print(history)
```

```python
loss, mae, mse = model.evaluate(test_df, test_labels, verbose=2)

print("Testing set Mean Abs Error: {:5.2f} expenses".format(mae))

test_predictions = model.predict(test_df).flatten()

a = plt.axes(aspect='equal')
plt.scatter(test_labels, test_predictions)
plt.xlabel('True values (expenses)')
plt.ylabel('Predictions (expenses)')
lims = [0, 50000]
plt.xlim(lims)
plt.ylim(lims)
_ = plt.plot(lims,lims)
```
